```python
# analysis of patient_age
min_age = min(df_train["patient_age"])
max_age = max(df_train["patient_age"])
print(f"Minimum age is {min_age}")
print(f"Maximum age is {max_age}")
less_than_30, less_than_40, less_than_50, less_than_60, less_than_70, less_than_80, less_than_90, less_than_100 = ([] for i in range(8))
for cell in df_train["patient_age"]:
 if cell >= 18 and cell < 30:
  less_than_30.append(cell)
 elif cell >= 30 and cell < 40:
  less_than_40.append(cell)
 elif cell >= 40 and cell < 50:
  less_than_50.append(cell)
 elif cell >= 50 and cell < 60:
  less_than_60.append(cell)
 elif cell >= 60 and cell < 70:
  less_than_70.append(cell)
 elif cell >= 70 and cell < 80:
  less_than_80.append(cell)
 elif cell >= 80 and cell < 90:
  less_than_90.append(cell)
 elif cell >= 90 and cell < 100:
  less_than_100.append(cell)
print(f"People aged 18-29: {len(less_than_30)}")
print(f"People aged 30-39: {len(less_than_40)}")
print(f"People aged 40-49: {len(less_than_50)}")
print(f"People aged 50-59: {len(less_than_60)}")
print(f"People aged 60-69: {len(less_than_70)}")
print(f"People aged 70-79: {len(less_than_80)}")
print(f"People aged 80-89: {len(less_than_90)}")
print(f"People aged 90-99: {len(less_than_100)}")
# results: people from ages 50-59 are diagnosed the most, closely followed by people from ages 60-69
plt.plot()

```
